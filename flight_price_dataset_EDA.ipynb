```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_excel("flight_price.xlsx")
df.head()
```

```python
df.info()
```

```python
df["Date"] = df["Date_of_Journey"].str.split("/").str[0].astype(int)
df["Month"] = df["Date_of_Journey"].str.split("/").str[1].astype(int)
df["Year"] = df["Date_of_Journey"].str.split("/").str[2].astype(int)

df.drop(columns="Date_of_Journey", inplace=True)

df.head()
```

```python
df["Arrival_Time"] = df["Arrival_Time"].str.split(" ").str[0]

df["Arrival_Hour"] = df["Arrival_Time"].str.split(":").str[0].astype(int)
df["Arrival_Min"] = df["Arrival_Time"].str.split(":").str[1].astype(int)

df.drop(columns="Arrival_Time", inplace=True)

df.head()
```

```python
df["Dep_Hour"] = df["Dep_Time"].str.split(":").str[0].astype(int)
df["Dep_Min"] = df["Dep_Time"].str.split(":").str[1].astype(int)

df.drop(columns="Dep_Time", inplace=True)

df.head()
```

```python
df["Duration_Hour"] = df["Duration"].str.extract(r"(\d+)h").fillna(0).astype(int)
df["Duration_Min"] = df["Duration"].str.extract(r"(\d+)m").fillna(0).astype(int)

df["Total_time_duration"] = df["Duration_Hour"] * 60 + df["Duration_Min"]

df.drop(columns=["Duration","Duration_Hour","Duration_Min"], inplace=True)

df.head()
```

```python
df["Total_Stops"] = df["Total_Stops"].fillna(df["Total_Stops"].mode()[0])
```

```python
df["Total_Stops"].value_counts()
```

```python
encoder = OrdinalEncoder(
    categories=[["non-stop", "1 stop", "2 stops", "3 stops", "4 stops"]]
)
encoded_array = encoder.fit_transform(df[["Total_Stops"]])

df["Total_Stops"] = encoded_array
df.head()
```

```python
df.info()
```

```python
df.drop(columns=["Route"],inplace=True)
df.head()
```

```python
df.info()
```

```python
encoder = OneHotEncoder(sparse_output=False)
encoded_array = encoder.fit_transform(
    df[
        [
            "Airline",
            "Source",
            "Destination",
            "Additional_Info",
        ]
    ]
)
# print(encoded_array)

encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoder.get_feature_names_out(),
    index=df.index,
)
```

```python
df = pd.concat(
    [
        df.drop(
            columns=["Airline", "Source", "Destination", "Additional_Info"],
        ),
        encoded_df,
    ],
    axis=1,
)
df.head()
```
